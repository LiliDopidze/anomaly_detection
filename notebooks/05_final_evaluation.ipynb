# 5. Locked final evaluation and frozen inference

Both are disabled by default. A qualified selection is required. Inference accepts a new canonical pack after the experiment window and never reads evaluation truth.

```python
from pathlib import Path
import json
import sys
import pandas as pd

ROOT = next(
    p for p in [Path.cwd(), *Path.cwd().parents]
    if (p / "pyproject.toml").exists()
)
sys.path.insert(0, str(ROOT / "src"))
from telco_anomaly.pipeline import runtime

POLICY = runtime(ROOT / "configs/pipeline.yml")
PACK = Path(POLICY["pack"])
RUN = Path(POLICY["run"])
```

```python
from telco_anomaly.pipeline import run_holdout, infer

OPEN_FINAL_TEST = False
if OPEN_FINAL_TEST:
    display(run_holdout(RUN))
else:
    print("Final test remains unopened.")
```

```python
RUN_INFERENCE = False
NEW_PACK = ROOT / "data/new_observations"
INFERENCE_OUTPUT = ROOT / "outputs/new_observations"
if RUN_INFERENCE:
    display(infer(RUN, NEW_PACK, INFERENCE_OUTPUT))
else:
    print("Inference is disabled until a qualified model and a new pack exist.")
```

The holdout ledger is written before truth is read and survives deletion of run outputs. Frozen code, inputs and model checksums are checked before scoring. A changed cadence requires recalibration. Filesystem separation prevents accidental label ingestion; it is not an operating-system permissions sandbox. Public-data integration, independent generator qualification and operator deployment remain later evidence gates.
